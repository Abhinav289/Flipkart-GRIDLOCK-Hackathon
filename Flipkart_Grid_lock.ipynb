```python
def parse_timestamp(ts):
    parts = str(ts).split(':')
    return int(parts[0]) + int(parts[1])/60

```

```python
trn['hour'] = trn['timestamp'].apply(parse_timestamp)
tst['hour'] = tst['timestamp'].apply(parse_timestamp)

print(trn['hour'].describe())
print(trn['hour'].unique()[:20])
print(tst['hour'].unique()[:20])
```

```python
trn['LargeVehicles'] = trn['LargeVehicles'].map({'Not Allowed': 0, 'Allowed': 1})
tst['LargeVehicles'] = tst['LargeVehicles'].map({'Not Allowed': 0, 'Allowed': 1})

trn['Landmarks'] = trn['Landmarks'].map({'No': 0, 'Yes': 1})
tst['Landmarks'] = tst['Landmarks'].map({'No': 0, 'Yes': 1})

print("Converted 'LargeVehicles' and 'Landmarks' columns in trn DataFrame:")
display(trn[['LargeVehicles', 'Landmarks']].head())

print("\nConverted 'LargeVehicles' and 'Landmarks' columns in tst DataFrame:")
display(tst[['LargeVehicles', 'Landmarks']].head())
```

```python
geohash_mode_trn = trn.groupby('geohash')['RoadType'].apply(lambda x: x.mode()[0] if not x.mode().empty else None)
geohash_mode_trn.nunique()
```

```python
trn['RoadType'] = trn['RoadType'].fillna(trn['geohash'].map(geohash_mode_trn))
```

```python
geohash_mode_tst = tst.groupby('geohash')['RoadType'].apply(lambda x: x.mode()[0] if not x.mode().empty else None)
geohash_mode_tst.nunique()
```

```python
tst['RoadType'] = tst['RoadType'].fillna(tst['geohash'].map(geohash_mode_tst))
```

```python
global_mode_trn = trn['RoadType'].mode()[0]
global_mode_tst = tst['RoadType'].mode()[0]
print(global_mode_trn,global_mode_tst)
```

```python
trn['RoadType'] = trn['RoadType'].fillna(global_mode_trn)
tst['RoadType'] = tst['RoadType'].fillna(global_mode_tst)
```

```python
print("Missing 'RoadType' values in trn after imputation:")
print(trn['RoadType'].isnull().sum())
print("\nMissing 'RoadType' values in tst after imputation:")
print(tst['RoadType'].isnull().sum())

print("\nFirst 5 rows of 'RoadType' in trn after imputation:")
display(trn[['geohash', 'RoadType']].head())

print("\nFirst 5 rows of 'RoadType' in tst after imputation:")
display(tst[['geohash', 'RoadType']].head())
```

```python
trn['Temperature'] = trn['Temperature'].fillna(trn.groupby(['day', 'hour'])['Temperature'].transform('median'))
trn
```

```python
trn['Temperature'] = trn['Temperature'].fillna(trn['Temperature'].median())
```

```python
tst['Temperature'] = tst['Temperature'].fillna(tst.groupby(['day', 'hour'])['Temperature'].transform('median'))
tst['Temperature'] = tst['Temperature'].fillna(tst['Temperature'].median())
```

```python

weather_mode_by_time_trn = trn.groupby(['day', 'hour'])['Weather'].apply(lambda x: x.mode()[0] if not x.mode().empty else None)
trn['Weather'] = trn.apply(lambda row: row['Weather'] if pd.notna(row['Weather']) else weather_mode_by_time_trn.get((row['day'], row['hour']), None), axis=1)

global_mode_weather_trn = trn['Weather'].mode()[0]
trn['Weather'] = trn['Weather'].fillna(global_mode_weather_trn)
```

```python
weather_mode_by_time_tst = tst.groupby(['day', 'hour'])['Weather'].apply(lambda x: x.mode()[0] if not x.mode().empty else None)
tst['Weather'] = tst.apply(lambda row: row['Weather'] if pd.notna(row['Weather']) else weather_mode_by_time_tst.get((row['day'], row['hour']), None), axis=1)

global_mode_weather_tst = tst['Weather'].mode()[0]
tst['Weather'] = tst['Weather'].fillna(global_mode_weather_tst)
```

```python
print("Missing 'Weather' values in trn after imputation:")
print(trn['Weather'].isnull().sum())
print("\nMissing 'Weather' values in tst after imputation:")
print(tst['Weather'].isnull().sum())

print("\nFirst 5 rows of 'Weather' in trn after imputation:")
display(trn[['day', 'hour', 'Weather']].head())

print("\nFirst 5 rows of 'Weather' in tst after imputation:")
display(tst[['day', 'hour', 'Weather']].head())
```

```python
geohash_demand_map = trn.groupby('geohash')['demand'].mean().to_dict()
global_mean_demand = trn['demand'].mean()
```

```python
trn['historical_demand'] = trn['geohash'].map(geohash_demand_map)
```

```python
tst['historical_demand'] = tst['geohash'].map(geohash_demand_map).fillna(global_mean_demand)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
categorical_cols = ['RoadType', 'Weather', 'geohash']

for col in categorical_cols:
    le = LabelEncoder()
    le.fit(list(trn[col].astype(str)) + list(tst[col].astype(str)))


    trn[col] = le.transform(trn[col].astype(str))
    tst[col] = le.transform(tst[col].astype(str))
```

```python

drop_cols = ['Index', 'timestamp', 'demand']
features = [col for col in trn.columns if col not in drop_cols]

X = trn[features]
y = trn['demand']

# Time-Based Split (Last 20% for validation)
split_idx = int(len(trn) * 0.8)

X_train, X_val = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_val = y.iloc[:split_idx], y.iloc[split_idx:]

print(f"Training data shape: {X_train.shape}")
print(f"Validation data shape: {X_val.shape}")
```

```python
from sklearn.metrics import r2_score
```

```python
from xgboost import XGBRegressor
```

```python
from skopt import BayesSearchCV
from skopt.space import Real, Integer
```

```python
xgb = XGBRegressor(random_state=42, n_jobs=-1)
```

```python
search_spaces = {
    'n_estimators': Integer(100,500),
    'learning_rate': Real(0.001, 0.1, prior='log-uniform'),
    'max_depth': Integer(1, 10),
    'subsample': Real(0.5, 1.0),
    'colsample_bytree': Real(0.5, 1.0),
    'min_child_weight': Integer(10, 20)
}
```

```python
from sklearn.model_selection import TimeSeriesSplit

tscv = TimeSeriesSplit(n_splits=3)
```

```python
bayes_search = BayesSearchCV(
    estimator=xgb,
    search_spaces=search_spaces,
    scoring='r2',          # We want to maximize the R2 score
    cv=tscv,               # Time-based cross validation
    n_iter=30,             # Number of parameter combinations to try (increase if you have time)
    n_jobs=-1,
    random_state=42,
    verbose=1
)
```

```python
print("Starting Bayesian Optimization....")
bayes_search.fit(X_train, y_train)
```

```python
best_xgb_model = bayes_search.best_estimator_
print(f"\nBest Parameters found: {bayes_search.best_params_}")
```

```python
val_preds = best_xgb_model.predict(X_val)
r2 = r2_score(y_val, val_preds)
hackathon_score = max(0, 100 * r2)
```

```python
print(f"\nValidation R2 Score (Holdout): {r2:.4f}")
print(f"Estimated Hackathon Score: {hackathon_score:.4f}")
```

```python
print("Retraining on full data...")
final_model = best_xgb_model # This already contains your best parameters
final_model.fit(X, y)
```

```python
X_test = tst[features]
```

```python
print("Generating predictions...")
test_predictions = final_model.predict(X_test)
```

```python
submission = pd.DataFrame({
    'Index': tst['Index'],
    'demand': test_predictions
})
```
